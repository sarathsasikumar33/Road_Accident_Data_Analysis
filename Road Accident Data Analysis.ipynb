# 1. Project Overview

This project focuses on analysing historical road accident data to understand patterns in accident severity, casualty rates, environmental conditions, road layouts, vehicle types, and geographic locations.

The project uses a public road accident dataset (Road Accident Dataset.xlsx) containing detailed information on individual collision incidents, including date, time, weather, road surface, light conditions, speed limits, casualty counts, and vehicle involvement.

Using Microsoft Excel, SQL, Python, and Power BI, the project will involve data cleaning, exploratory data analysis, severity risk factor analysis, spatial-temporal analysis, and dashboard development.

The main aim is to transform raw collision logs into meaningful insights that can help urban planners, traffic management authorities, and safety officers identify high-risk conditions and make data-driven infrastructure and enforcement decisions.

# 2. Business Problem

Traffic safety administrators, municipal authorities, and logistics managers often struggle to identify which combination of environmental factors (e.g., weather, lighting, road surface) and infrastructure conditions (e.g., speed limits, junction controls, road types) contribute most significantly to severe or fatal collisions.

Although collision reports capture rich incident details, the data is typically high-volume and multi-dimensional. Without systematic analysis, identifying high-risk collision patterns, critical road hazard correlations, or temporal spikes (such as rush-hour vs. nighttime risks) remains difficult.

This project addresses this issue by analysing structured road accident data to examine severity distributions, identify high-frequency hazard combinations, evaluate casualty rates across vehicle categories, and compare urban versus rural collision dynamics.

The analysis will provide descriptive and diagnostic insights from available accident records. It will not establish legal fault or driver-level causation, as the dataset does not contain driver toxicology reports or qualitative police investigation text.

# 4.1 Important Dataset Columns

Accident_Index — Unique identifier for each collision record.

Accident Date — Date on which the collision occurred.

Day_of_Week — Day of the week when the incident took place.

Time — Recorded time of day for the crash.

Local_Authority_(District) — Municipal district responsible for the accident site.

Police_Force — Reporting police force jurisdiction.

Latitude & Longitude — Exact spatial coordinates of the collision.

Accident_Severity — Severity classification (Fatal, Serious, Slight).

Number_of_Casualties — Count of total injuries or fatalities recorded.

Number_of_Vehicles — Count of motor vehicles involved.

Speed_limit — Posted speed limit at the collision location.

Road_Type — Infrastructure type (e.g., Single carriageway, Dual carriageway, Roundabout).

Junction_Control & Junction_Detail — Type of junction and control mechanism present.

Light_Conditions — Illumination levels during the accident.

Weather_Conditions — Weather conditions recorded at the scene.

Road_Surface_Conditions — Road conditions (e.g., Dry, Wet/Damp, Snow, Ice).

Special_Conditions_at_Site — Site hazards (e.g., Traffic signal out, Oil/Diesel spill).

Carriageway_Hazards — Recorded obstacles on the carriageway.

Urban_or_Rural_Area — Regional classification (Urban vs. Rural).

Vehicle_Type — Class of primary vehicle involved (e.g., Car, Motorcycle, Bus, Goods Vehicle).

# 6. Scope and Limitations

### 6.1 Scope

Analysis of historical accident logs provided in Road Accident Dataset.xlsx.

Descriptive and exploratory analysis using Python (Pandas, NumPy, Matplotlib, Seaborn), SQL, and Excel.

Geospatial mapping utilizing recorded Latitude and Longitude values.

Dashboard creation and visualization with Power BI.

### 6.2 Limitations

Reporting bias: The dataset reflects formally reported police incidents; minor unreported collisions are absent.

Missing data: Optional fields (e.g., specific carriageway hazards) contain high proportions of missing/null values, requiring clear sample reporting.

Lack of driver demographics: Human attributes such as driver age, experience, seatbelt usage, or impairment levels are not included.

Correlation vs. Causation: Findings represent statistical associations between conditions and crash outcomes, not definitive legal fault.

Urban/Rural sample skew: Absolute collision counts are higher in urban areas, requiring proportion-based metrics for valid comparisons.
